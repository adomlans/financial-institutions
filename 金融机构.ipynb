```python
Day3 利率风险管理
今日目标：掌握银行和固收组合最核心的利率风险管理工具——久期缺口与重定价缺口，能算清利率变动对资产负债表和净利息收入的影响。
知识点模块：
- 久期缺口（DGAP）：DGAP = 资产久期 −（负债/资产）× 负债久期。它衡量资产与负债对利率敏感性的错配程度。利率上升时资产和负债价值同时下跌，若资产久期更长（正缺口），资产跌幅更大，净权益受损；权益变动近似为 −DGAP × Δr/(1+r) × 资产规模。
- 重定价缺口（GAP）：GAP = 利率敏感资产 − 利率敏感负债。它衡量短期内重定价的现金流错配：正缺口下利率上升→净利息收入增加；负缺口下利率下降→净利息收入增加。久期缺口管"存量价值"，重定价缺口管"流量收益"，两者互补使用。
- 管理手段三层：缺口匹配（让资产与负债的久期/重定价期限对齐）、产品结构调整、利率衍生品对冲（利率互换、债券期货、利率期权）。实务中银行多靠衍生品做主动对冲，因为调整存贷结构成本高、周期长。


自检表（验收项与参考答案）：
验收项
参考答案
久期缺口为正时利率上升对权益的影响
资产价值跌幅大于负债价值跌幅，权益下降；DGAP=DA−k·DL为正即资产久期长于经杠杆调整的负债久期
重定价缺口（GAP）的盈亏方向
GAP=RSA−RSL；正缺口下利率上升→净利息收入增加；负缺口下利率下降→净利息收入增加
银行管理利率风险的三个手段
缺口匹配、产品结构调整、利率衍生品（互换/债券期货/期权）对冲
```

```python
#Day3
# 银行久期缺口与利率风险（Day3）
A = 100.0    # 总资产(亿元)
L = 90.0     # 总负债(亿元)
DA = 5.0     # 资产组合久期(年)
DL = 2.0     # 负债组合久期(年)
k = L / A
DGAP = DA - k * DL                 # 久期缺口
dE = -DGAP * 0.01 / 1.05 * A       # 利率+1%时权益变动
print('久期缺口 DGAP:', round(DGAP, 3))
print('利率上升1个百分点, 权益变动(亿元):', round(dE, 3))
RSA, RSL = 60.0, 40.0              # 利率敏感资产/负债
GAP = RSA - RSL
dNII = GAP * 0.01
print('重定价缺口:', GAP, '→ 利率+1%净利息收入变动(亿元):', dNII)
```

```python
Day5 中央银行与货币政策
今日目标：看懂央行资产负债表与三大工具，会算货币乘数，分清数量型与价格型调控。
知识点模块：
- 央行资产负债表：资产端主要是外汇储备、对政府债权、对其他存款性公司债权（逆回购、MLF、再贷款等投放）；负债端主要是货币发行（现金）、存款准备金、政府存款。资产扩张=基础货币投放，缩表=回收。
- 三大传统工具：存款准备金率（调节乘数）、再贴现与再贷款（向银行提供流动性）、公开市场操作（买卖债券，最常用最灵活）。加上结构性工具（支农支小再贷款、PSL等）形成中国政策工具箱。
- 存款创造与货币乘数：银行体系"多存多贷"：一笔存款被反复贷放形成派生存款，最大存款=初始存款/准备金率；考虑现金漏损与超额准备金后乘数为 (1+c)/(r+c+e)。降准通过放大乘数扩张信用，不改变基础货币。
- 数量型 vs 价格型：数量型直接调控量（准备金率、信贷额度）；价格型调控利率（政策利率、LPR）。中国正从数量为主转向价格为主，利率走廊（7天逆回购为中枢）逐步成为主锚。




自检表（验收项与参考答案）：
验收项
参考答案
三大传统货币政策工具
存款准备金率、再贴现（再贷款）、公开市场操作
降准如何影响货币供给
不改变基础货币，但提高货币乘数→同等基础货币创造更多存款与信贷
央行买入国债与降准的区别
前者直接投放基础货币并扩表（QE式）；后者放大乘数但不改变基础货币
```

```python
Day6 政策利率传导
今日目标：把"政策利率→市场利率→贷款利率"这条链走通，理解 LPR 机制与中美政策框架差异，看懂 QE 与缩表。
知识点模块：
- 中国传导链：7天逆回购（政策利率）→ DR007（货币市场利率）→ 银行负债成本 → LPR（贷款市场报价利率）→ 实体融资成本。政策利率动 10bp，传导到贷款利率需要时间且并非等量，关键在银行负债成本与加点。
- LPR 机制：LPR = MLF 利率 + 报价行加点，由18家报价行每月报价形成；1年期LPR是企业贷款锚，5年期以上LPR是房贷锚。降 MLF 是降 LPR 的主要工具，但加点受银行负债成本约束，降MLF并不必然等幅传导。
- 美联储框架：以联邦基金利率目标区间为锚，辅以准备金利率（IOER）与隔夜逆回购（ON RRP）构筑利率走廊；量化宽松（QE）通过买入国债与MBS扩表压低长端利率，量化紧缩（QT）通过到期不续缩表回收流动性。
- 传导失效情形：流动性陷阱（利率已近零、货币需求弹性无限，降息无效）与信用收缩（银行惜贷、企业不愿借贷，宽松政策滞留在银行体系）。这是判断"政策失灵"的两种经典场景。


自检表（验收项与参考答案）：
验收项
参考答案
政策利率到实体融资的传导链
7天逆回购→DR007→银行负债成本→LPR→贷款利率→实体融资需求
LPR 由什么构成
LPR=MLF+报价行加点；加点反映负债成本、风险与利润诉求
QE 与 QT 的方向
QE 央行购入资产扩表投放流动性；QT 到期不续或卖出资产缩表回收流动性
传导失效的两种情形
流动性陷阱（利率极低时降息无效）与信用收缩（银行惜贷、实体不愿借贷）
```

```python
阶段三 债券与股票市场（Day7–9）
Day7 债券市场
今日目标：看懂中国债券市场结构，会分解信用利差，理解利率债与信用债的分析差异。
知识点模块：
- 债券类型分层：利率债（国债、地方政府债、政策性金融债）以政府信用背书、几乎无违约风险，定价锚是利率；信用债（企业债、公司债、中期票据、短融）含发行人信用风险，定价锚是"利率+信用利差"；可转债是"债券+股票期权"的混合体，有债底保护。
- 市场结构：中国债券市场以银行间市场为主（机构投资者、托管于中债登与上清所、占存量约九成），交易所市场为辅（含散户、托管于中登）。两个市场在投资者准入、质押回购、做市规则上存在差异，存在跨市场价差套利机会。
- 信用利差分解：信用利差=企业债收益率−同期限国债收益率=流动性溢价+信用风险溢价+税收因素。流动性溢价补偿交易不活跃，信用风险溢价补偿违约风险，税收因素源于国债免税。经济下行时违约预期上升，利差走阔，是信用周期的领先信号。
- 发行与交易：一级市场发行以招标/簿记建档定价；二级市场以做市商双边报价成交。质押式回购是债市加杠杆的主要渠道，也放大利率波动。


自检表（验收项与参考答案）：
验收项
参考答案
信用利差的定义与拆解
企业债收益率−同期限国债收益率=流动性溢价+信用风险溢价+税收因素
经济下行时信用利差走势
走阔：违约预期上升、流动性收紧、风险厌恶升温
银行间与交易所市场差异
投资者准入（机构为主vs含零售）、托管结算（中债登/上清所vs中登）、质押与做市规则不同
```

```python
Day9 市场微观结构与流动性
今日目标：看懂订单簿与价差，会算冲击成本，理解高频交易对市场质量的双面影响。
知识点模块：
- 订单簿：限价单挂在买/卖两侧形成盘口，市价单按对手价成交。最佳买价与最佳卖价之差即买卖价差（spread），是流动性最直接的度量：价差窄→交易成本低、流动性好。
- 市场深度与冲击成本：深度指不显著改变价格的前提下市场能吸收的订单量。大单按档位逐层消耗深度，平均成交价偏离盘口价的部分即冲击成本。深度不足时大单引发价格滑点，这是机构交易最关心的成本之一。
- 流动性分层：核心资产（大盘股、国债）深度好、价差窄；小盘股与信用债流动性分层明显，交易需分批或使用算法拆单。
- 高频交易双面性：做市型高频收窄价差、提供流动性，改善市场质量；抢跑、幌骗、操纵型行为损害公平、放大波动，是监管与《暗池》等书籍争议的焦点。


自检表（验收项与参考答案）：
验收项
参考答案
买卖价差代表什么
做市成本与流动性水平；价差窄→交易成本低、流动性好
市场深度的含义
不显著影响价格的前提下市场能吸收的订单量；深度不足→大单冲击成本高
高频交易对市场质量的双面影响
做市收窄价差、提供流动性；抢跑与操纵行为损害公平与信心
```

```python
#Day9
# 订单簿与冲击成本（Day9）
best_bid, best_ask = 10.00, 10.02
spread = best_ask - best_bid
depth_ask = 4000                    # 卖盘档位深度(股)
buy_qty = 10000                     # 市价买入量
if buy_qty <= depth_ask:
    impact = spread / 2
else:
    impact = spread / 2 + (buy_qty - depth_ask) / depth_ask * 0.01
print('买卖价差:', spread, '卖一深度:', depth_ask)
print('市价买入{}股的平均冲击成本(元/股): {:.4f}'.format(buy_qty, impact))
print('价差窄→流动性好; 深度不足→大单冲击成本高')
```

```python
Day12 影子银行与系统性风险
今日目标：识别影子银行链条与风险来源，理解宏观审慎（MPA）为何存在，看懂同业传染机制。
知识点模块：
- 影子银行：游离于传统银行监管之外的信用中介，中国形态主要是银行表外业务（同业理财、通道业务、非标投资）与信托、P2P等。典型链条：银行发行同业存单→购买同业理财→委外给非银→加杠杆买债/投非标，层层放大。
- 风险来源：期限错配（短资金投长资产）、流动性错配（可随时赎回的钱投了不可变现资产）、杠杆叠加、监管套利（规避资本与信贷约束）、信息不透明。2017–2018金融去杠杆针对的就是这条链，资管新规与同业负债纳入MPA考核切断套利。
- 宏观审慎（MPA）：宏观审慎评估体系约束广义信贷增速与杠杆，逆周期调节，防止信用过快扩张与系统性风险积累，与微观审慎（单家机构风险）互补。
- 传染机制：机构间同业互持与共同资产持有，使单一机构违约/赎回潮通过敞口传导为连锁反应（P2P暴雷与2013钱荒是典型案例），这是"系统性风险"区别于"个别风险"的本质。



自检表（验收项与参考答案）：
验收项
参考答案
影子银行的主要风险来源
期限错配、流动性错配、高杠杆、监管套利、信息不透明
MPA 的作用
宏观审慎评估：约束广义信贷增速与杠杆，防止信用过快扩张积累系统性风险
同业链条如何传染
机构间同业互持与共同资产→单一机构违约/赎回→对手方流动性收紧→抛售与连锁违约G
```

```python
Day13 外汇市场与汇率
今日目标：看懂即期/远期汇率与利率平价，理解人民币汇率制度与利差对汇率的影响。
知识点模块：
- 外汇市场基础：即期汇率（当前交割）、远期汇率（约定未来交割）、掉期（即期+远期组合）。直接标价法下 1 美元兑多少人民币（USD/CNY），数值上升=人民币贬值。
- 汇率决定：长期看购买力平价（两国物价水平比）；短期看利率平价（利差驱动资本流动）与国际收支（贸易顺差→本币升值压力）。利率平价：高利率货币远期贴水（预期贬值），低利率货币远期升水（预期升值）；市场远期偏离理论值即存在套利空间。
- 人民币汇率制度：以市场供求为基础、参考一篮子货币、有管理的浮动汇率制度；央行通过中间价设定与逆周期因子平滑过度波动，外汇储备用于干预与稳定预期。
- 利差与资本流动：中美利差倒挂（人民币利率低于美元）→套息资金流出→人民币贬值压力；反之利差走阔→升值压力。汇率是宏观流动性的"总开关"之一，与国内资产价格联动。



检表（验收项与参考答案）：
验收项
参考答案
利率平价的核心结论
高利率货币远期贴水、低利率货币远期升水；利差被远期汇率消化
中美利差倒挂对人民币汇率的影响
套息资金流出→人民币贬值压力；央行通过中间价与工具平滑波动
远期汇率理论值公式
F=S×(1+本币利率)/(1+外币利率)；市场价偏离理论值即有套利空间
```

```python
# 利率平价与远期汇率（Day13）
S = 7.20         # 即期汇率 美元/人民币
r_cny, r_usd = 0.030, 0.050
F_theory = S * (1 + r_cny) / (1 + r_usd)
print('利率平价理论远期汇率: {:.4f}'.format(F_theory))
print('人民币利率低 → 人民币远期升水(远期升值) → F < S')
F_market = 7.15
print('市场远期', F_market, '与理论值差:', round(F_market - F_theory, 4))
print('偏差即套利空间: 低息货币借入 → 高息货币投放 → 远期锁汇')
```

```python
Day15 金融风险与危机
今日目标：把十五天知识串成危机视角：看懂危机如何传导、为什么流动性危机会变信用危机，以及监管如何应对。
知识点模块：
- 危机类型：流动性危机（资产卖不出去/资金借不到，但基本面未坏）、信用危机（违约与偿付能力恶化）、货币危机（汇率崩溃）、债务危机（主权/企业无力偿债）。现实中多类型叠加。
- 统一传导链条：资产价格下跌→抵押品减值→去杠杆抛售→价格进一步下跌（负反馈）→信贷收缩→实体衰退。2008年：次贷违约→MBS/CDO 价格崩溃→雷曼倒闭→同业传染→全球信贷冻结。
- 流动性危机→信用危机：被迫抛售压低资产价格→杠杆机构爆仓违约→对手方损失→偿付能力危机。这就是为什么央行要当"最后贷款人"：在恐慌中提供紧急流动性，阻断自我实现的挤兑与抛售。
- 预警与监测：盯杠杆率、信用利差（利差骤扩是恐慌信号）、房价、经常账户、银行同业拆借利率（如2013年钱荒）。压力测试（宏观+微观）用于评估极端情景下的资本缺口。


自检表（验收项与参考答案）：
验收项
参考答案
危机的典型传导链条
资产价格下跌→抵押品减值→去杠杆抛售→价格再跌→信贷收缩→实体衰退
流动性危机如何演化为信用危机
被迫抛售压低价格→杠杆机构爆仓违约→对手方损失→偿付能力危机
最后贷款人的作用
央行在恐慌中提供紧急流动性，阻断挤兑与抛售的自我实现
```

```python
# 银行间网络传染模拟（Day15）
import numpy as np
rng = np.random.default_rng(15)
n = 6
exposure = rng.uniform(0, 5, (n, n))   # 同业敞口矩阵(模拟)
np.fill_diagonal(exposure, 0)
capital = rng.uniform(4, 10, n)
defaulted = np.zeros(n, dtype=bool)
defaulted[0] = True
rounds = 0
for rnd in range(3):
    losses = exposure[:, defaulted].sum(axis=1)
    new_default = (capital - losses) < 0
    if not (new_default & ~defaulted).any():
        rounds = rnd
        break
    defaulted |= new_default
    rounds = rnd + 1
print('传染轮数:', rounds, '最终违约银行数: {} / {}'.format(defaulted.sum(), n))
print('一家违约 → 同业敞口传递 → 连锁违约(流动性+信用危机)')
```
